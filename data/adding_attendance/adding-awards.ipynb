This notebook adds awards such as rosie finalists to a list of students

```python
import pandas as pd
import os
```

```python
user_data_path = os.path.join(os.getcwd(), "..", "User_Datas.csv") # UPDATE NAME ACCORDINGLY
if not os.path.exists(user_data_path):
    raise FileNotFoundError(f"User_Data.csv not found at {user_data_path}")
user_data = pd.read_csv(user_data_path)

# Function to add awards to users
def add_awards_to_users(names_list: list[str], award_name: str, user_data: pd.DataFrame) -> pd.DataFrame:
    """
    Add an award to specified users in the user_data DataFrame.
    
    Args:
        names_list: List of user names to add the award to
        award_name: Name of the award to add
        user_data: DataFrame containing user data
        
    Returns:
        Updated DataFrame with awards added
    """
    for name in names_list:
        name = name.strip().title()
        
        # Check if user exists in the DataFrame
        name_list = user_data['User'].apply(lambda x: x.strip().title()).tolist()
        if name not in name_list:
            print(f'WARNING: {name} is not in the user_data df')
            # Skip adding users not in the CSV
            continue
        else:
            # Add award to existing user
            index = name_list.index(name)
            # If they already have awards, append the new award
            if pd.notna(user_data.loc[index, 'Awards']) and user_data.loc[index, 'Awards'] != '':
                user_data.loc[index, 'Awards'] += f'|{award_name}'
            # If they don't have any awards yet, add the new award
            else:
                user_data.loc[index, 'Awards'] = award_name
    
    return user_data

award = "2025 Researcher" # DO NOT FORGET TO CHANGE THIS!
names = [
    "Michael Wood",
    "Andrew Needham",
    "Will Lassiter",
    "Alex Ewart",
    "Mason Beynon",
    "Zoe Kirkman",
    "Adam Swedlund",
    "Preston Whitcomb",
    "Jacob Tyree",
    "Evan Schubert",
    "Mikhail Filippov",
    "Tanner Carlson",
    "Franklin Cole",
    "Ryan Pal Hilgendorf",
    "Autumn Mizer",
    "Ben Williams",
    "Levi Wolfgramm",
    "Excel Olatunji",
    "Muhammad Alliyu",
    "Alhagie Boye",
    "Dominic Zagra",
    "Vincenzo Bean",
    "Joseph LoDuca",
    "Eli Vang",
    "Xander Ede",
    "Joseph Beattie",
    "Alex Neher",
    "Gavin Childress",
    "Josh Myers",
    "Zach Peterson",
    "Peter Kwaterski"
]

user_data = add_awards_to_users(names, award, user_data)
user_data.to_csv(user_data_path, index=False)
```
